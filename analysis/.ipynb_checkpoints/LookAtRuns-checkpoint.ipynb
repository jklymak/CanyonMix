```python
import xmitgcm as xm
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
%matplotlib ipympl
import matplotlib.colors as mcolors
from xmitgcm.file_utils import clear_cache
clear_cache()


with xm.open_mdsdataset('../results/StraightSlopeUpsidedownStrat003Dye001a/input/', iters='all', endian='<', prefix='spinup', ref_date='2020-01-01') as ds0:
    display(ds0)
    ds=ds0.isel(j=0, j_g=0).isel(time=48*2)
    ds00=ds0.isel(j=0, j_g=0).isel(time=0)
    fig, ax = plt.subplots(figsize=(6, 3),layout='constrained')
    #ax.pcolormesh(ds.XG, ds.Z, ds['UVEL'], clim=[-0.3, 0.3], #cmap='RdBu_r')
    pc = ax.pcolormesh(ds.XG, ds.Z, ds['SALT'].rolling(i=1).mean()-35.0, cmap='Greens', norm=mcolors.LogNorm(vmin=1e-2, vmax=100))
    ax.contour(ds.XC, ds.Z, ds['THETA'], levels=np.arange(27, 28, 0.02), colors='k')
    ax.contour(ds.XC, ds.Z, ds00['THETA'], levels=np.arange(27, 28, 0.02), colors='0.5')
    ax.contour(ds.XC, ds.Z, ds00['SALT'], levels=np.array([35, 100]), colors='k')
    #ax.set_xlim([90_000, 120_000])
    ax.set_title(f'{ds.time.values}')
    fig.colorbar(pc)
```

```python

todo = ['../results/StraightSlopeExpStrat004K001/input/', '../results/StraightSlopeExpStrat004K01/input/', '../results/StraightSlopeExpStrat004K03/input/',
        '../results/StraightSlopeExpStrat004K1/input/']
Ks = [1e-3, 1e-2, 3e-1, 1]
fig, axs = plt.subplots(4, 2, figsize=(8, 8), layout='constrained')
for nn, td in enumerate(todo):
    with (xm.open_mdsdataset(td, iters='all', endian='<', prefix='spinup', ref_date='2020-01-01') as ds0):
        ds=ds0.isel(j=0, j_g=0).isel(time=96)
        ds['THETA'] = ds['THETA'].where(ds['THETA'] > 0)
        ds['XG'] = ds['XG'] / 1e3
        ds['XC'] = ds['XC'] / 1e3

        ax = axs[nn, 0]
        pc0 = ax.pcolormesh(ds.XG, ds.Z, ds['UVEL'], cmap='RdBu_r', clim=(-0.1, 0.1))
        ax.contour(ds.XC, ds.Z, ds['THETA'], levels=np.arange(27, 28, 0.02), colors='k')
        #ax.set_xlim([90_000, 120_000])
        ax.set_title(f'K = {Ks[nn]:.1e}')
        ax.set(facecolor='0.5')

        ax = axs[nn, 1]
        pc1 = ax.pcolormesh(ds.XG, ds.Z , ds['UVEL'] * (3e-1/Ks[nn])**(0.5), cmap='RdBu_r', clim=(-0.1, 0.1))
        ax.contour(ds.XC, ds.Z, ds['THETA'], levels=np.arange(27, 28, 0.02), colors='k')
        #ax.set_xlim([90_000, 120_000])
        ax.set_title(f'K = {Ks[nn]:.1e}')
        ax.set(xlim=[115, 135], ylim=[-1200, -500], facecolor='0.5')

fig.colorbar(pc0, ax=axs[:, 0], shrink=0.3, extend='both', label = 'U')
fig.colorbar(pc1, ax=axs[:, 1], shrink=0.3, extend='both', label='(3e-1/K) U' )
fig.savefig('Ks.png', dpi=200)
```
